```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\Customers.csv")
churn_status = pd.read_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\churn_status.csv")
usage_data = pd.read_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\usage_data.csv")
complaints = pd.read_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\complaints.csv")
```

```python
print(customers.shape)
print(churn_status.shape)
print(usage_data.shape)
print(complaints.shape)
```

# Customers Table — Null & Duplicate Check

```python
print("Nulls in customers:")
print(customers.isnull().sum())
print()
print("Duplicate customer_ids:", customers['customer_id'].duplicated().sum())
print()
print(customers.dtypes)
```

# Churn_status Table — Null & Duplicate Check

```python
print("Nulls in churn_status:")
print(churn_status.isnull().sum())
print()
print("Non-churned customers count:", (churn_status['is_churned']==0).sum())
print()
print("Duplicate customer_ids:", churn_status['customer_id'].duplicated().sum())
print()
print(churn_status.dtypes)
```

```python
churn_status['churn_date'] = pd.to_datetime(churn_status['churn_date'])
print(churn_status.dtypes)
```

# Usage_data — Null, Duplicate & Dtype Check

```python
print("Nulls in usage_data:")
print(usage_data.isnull().sum())
print()
print("Duplicate rows (customer_id + month):", usage_data.duplicated(subset=['customer_id', 'month']).sum())
print()
print(usage_data.dtypes)
```

```python
usage_data['month'] = pd.to_datetime(usage_data['month'])
print(usage_data.dtypes)
```

# Complaints — Null, Duplicate & Dtype Check

```python
print("Nulls in complaints:")
print(complaints.isnull().sum())
print()
print("Duplicate complaint_ids:", complaints['complaint_id'].duplicated().sum())
print()
print(complaints.dtypes)
```

```python
complaints['complaint_date'] = pd.to_datetime(complaints['complaint_date'])
print(complaints.dtypes)
```

```python
usage_orphans = ~usage_data['customer_id'].isin(customers['customer_id'])
print("Orphan records in usage_data:", usage_orphans.sum())

complaints_orphans = ~complaints['customer_id'].isin(customers['customer_id'])
print("Orphan records in complaints:", complaints_orphans.sum())

churn_orphans = ~churn_status['customer_id'].isin(customers['customer_id'])
print("Orphan records in churn_status:", churn_orphans.sum())
```

```python
customers.to_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\cleaned_dataset\customers_clean.csv", index=False)
churn_status.to_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\cleaned_dataset\churn_status_clean.csv", index=False)
usage_data.to_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\cleaned_dataset\usage_data_clean.csv", index=False)
complaints.to_csv(r"C:\Data Analytics\Projects\END TO END\TeleCom_Analysis\Dataset\cleaned_dataset\complaints_clean.csv", index=False)

print("All files saved successfully!")
```
